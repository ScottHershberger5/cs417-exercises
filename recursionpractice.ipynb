```python
def countdown_bad(n):
  if n <= 0:
    print('Blastoff')
  else:
    print(n)
    countdown_bad(n)

# countdown_bad(5)
```

```python
def python_recursion_eg( str ):
  if len(str) == 0:
    return 0
  count = 0

  if str[0] == "A":
    count = 1
  return python_recursion_eg(str[1:]) + count
python_recursion_eg("AlphabAt")
```
