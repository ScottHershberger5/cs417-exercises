```python
import requests

url = "https://api.open-meteo.com/v1/forecast"
params = {
    "latitude": 60,
    "longitude": -23,
    "current_weather": True
}

response = requests.get(url, params=params) #HTTP requests the data from the HTTP server​
data = response.json() #turns the response into a json file​ (dictonary)

url2 = "https://api.open-meteo.com/v1/forecast"
params = {
    "latitude": 40,
    "longitude": 55,
    "current_weather": True
}


response2 = requests.get(url2, params=params)
data2 = response2.json()
# print(data)
# print(data2)

print(f"location 1: lat = {data['latitude']} , long = {data['longitude']}")
print(f"location 2: lat = {data2['latitude']} , long = {data2['longitude']}")
print()
temp_diff = data['current_weather']['temperature'] - data2['current_weather']['temperature']
print(f"The temperature difference between location 1 and location 2 is {temp_diff} degrees.")
```
