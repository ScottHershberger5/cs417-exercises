```python
list_unsorted = [2,6,9,1,50,3,4,22,21,47]

list_sorted = [1,3,4,6,9,13,23,25,26]

def seq_search(list,item):
  pos = 0
  while pos < len(list):
    if list[pos] == item:
      return True
    pos += 1
  return False

def ord_seq_search(list,item):
  pos = 0
  while pos < len(list):
    if list[pos] == item:
      return True
    if list[pos] > item: #optimization
      return False
    pos += 1
  return False

def binary_search(list,item):
  start = 0
  last = len(list)
  while last - start > 1:
    mid = (start + last) // 2
    if mid == item:
      return True
    elif mid > item:
      last = mid - 1
    else:
      start = mid + 1


  if start == item: #checks the last two items
    return True
  elif last == item:
    return True
  else:
    return False
```

```python
seq_search(list_unsorted, 22) # should return true
ord_seq_search(list_sorted, 9) # should return true
seq_search(list_unsorted, 1000) # should return false
ord_seq_search(list_sorted, 1000) # should return false
binary_search(list_sorted, 3)#true
binary_search(list_sorted, 26)#true
# binary_search(list_sorted, 1000) #false
# binary_search(list_sorted, 2) #false
```
